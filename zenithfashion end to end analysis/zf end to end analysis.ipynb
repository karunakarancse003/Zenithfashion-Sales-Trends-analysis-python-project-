```python
import pandas as pd
import matplotlib.pyplot as plt

# ==========================================
# 1. DATE FORMAT
# ==========================================

df["Order_Date"] = pd.to_datetime(df["Order_Date"])


# ==========================================
# 2. MONTHLY SALES
# ==========================================

monthly_sales = (
    df.groupby(df["Order_Date"].dt.to_period("M"))["Sales_Amount"]
    .sum()
    .sort_index()
)


# ==========================================
# 3. APR 2023 - MAR 2024
# ==========================================

monthly_sales = monthly_sales[
    (monthly_sales.index >= "2023-04") &
    (monthly_sales.index <= "2024-03")
]


# ==========================================
# 4. CONVERT TO ₹ LAKH
# ==========================================

sales_lakh = monthly_sales / 100000


# ==========================================
# 5. MONTH LABELS
# ==========================================

months = sales_lakh.index.strftime("%b'%y")


# ==========================================
# 6. CREATE CHART
# ==========================================

fig, ax = plt.subplots(figsize=(13, 8))

x = list(range(len(months)))


# ==========================================
# 7. SALES LINE
# ==========================================

ax.plot(
    x,
    sales_lakh.values,
    color="#722C4A",
    linewidth=3,
    marker="o",
    markersize=6
)


# ==========================================
# 8. X-AXIS
# ==========================================

ax.set_xticks(x)

ax.set_xticklabels(
    months,
    fontsize=10,
    color="#46152B"
)

# X-axis TITLE
ax.set_xlabel(
    "Month",
    fontsize=12,
    color="#46152B",
    labelpad=10
)


# ==========================================
# 9. Y-AXIS
# ==========================================

ax.set_ylim(0, 5)

ax.set_yticks([
    0, 0.5, 1, 1.5, 2,
    2.5, 3, 3.5, 4, 4.5, 5
])

ax.set_yticklabels(
    [
        "0", "0.5", "1", "1.5", "2",
        "2.5", "3", "3.5", "4", "4.5", "5"
    ],
    fontsize=10,
    color="#46152B"
)

# Y-axis TITLE
ax.set_ylabel(
    "₹ Lakh",
    fontsize=12,
    color="#46152B",
    labelpad=10
)


# ==========================================
# 10. VALUE LABELS
# ==========================================

for i, value in enumerate(sales_lakh.values):

    ax.annotate(
        f"₹{value:.2f}L",
        (i, value),
        xytext=(0, 10),
        textcoords="offset points",
        ha="center",
        fontsize=9,
        color="#46152B"
    )


# ==========================================
# 11. MAIN TITLE
# ==========================================

ax.set_title(
    "Monthly Sales Trend (Apr 2023 – Mar 2024)",
    fontsize=22,
    fontweight="bold",
    color="#46152B",
    pad=25
)


# ==========================================
# 12. GRID
# ==========================================

ax.grid(
    axis="y",
    linestyle="-",
    alpha=0.25
)

ax.set_axisbelow(True)


# ==========================================
# 13. CLEAN LOOK
# ==========================================

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)


# ==========================================
# 14. SHOW
# ==========================================

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# =========================================================
# 1. DATE CONVERSION
# =========================================================

df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    errors="coerce"
)

# =========================================================
# 2. SALES AMOUNT → NUMERIC
# =========================================================

df["Sales_Amount"] = (
    df["Sales_Amount"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("₹", "", regex=False)
    .str.strip()
)

df["Sales_Amount"] = pd.to_numeric(
    df["Sales_Amount"],
    errors="coerce"
).fillna(0)

# =========================================================
# 3. CLEAN RETURN FLAG
# =========================================================

df["Return_Flag"] = (
    df["Return_Flag"]
    .astype("string")
    .str.strip()
    .str.upper()
)

# Check Return Flag
print("Return Flag Count:")
print(df["Return_Flag"].value_counts(dropna=False))

# =========================================================
# 4. CREATE QUARTER
# =========================================================

df["Quarter"] = (
    df["Order_Date"]
    .dt.to_period("Q")
    .astype(str)
)

# =========================================================
# 5. NORMAL SALES
# NO = Sales
# =========================================================

df["Normal_Sales"] = df["Sales_Amount"].where(
    df["Return_Flag"] == "NO",
    0
)

# =========================================================
# 6. RETURN SALES AMOUNT
# YES = Returns
# =========================================================

df["Return_Sales"] = df["Sales_Amount"].where(
    df["Return_Flag"] == "YES",
    0
)

# =========================================================
# 7. GROUP SALES
# =========================================================

sales_data = df.groupby(
    ["Quarter", "Category"]
)["Normal_Sales"].sum().reset_index()

# =========================================================
# 8. GROUP RETURNS
# =========================================================

return_data = df.groupby(
    ["Quarter", "Category"]
)["Return_Sales"].sum().reset_index()

# =========================================================
# 9. PIVOT SALES
# =========================================================

sales = sales_data.pivot(
    index="Quarter",
    columns="Category",
    values="Normal_Sales"
).fillna(0)

# =========================================================
# 10. PIVOT RETURNS
# =========================================================

returns = return_data.pivot(
    index="Quarter",
    columns="Category",
    values="Return_Sales"
).fillna(0)

# Make same order
returns = returns.reindex(
    index=sales.index,
    columns=sales.columns,
    fill_value=0
)

# =========================================================
# 11. QUARTERS & CATEGORIES
# =========================================================

quarters = list(sales.index)
categories = list(sales.columns)

print("\nCategories:")
print(categories)

# =========================================================
# 12. CATEGORY COLORS
# =========================================================

category_colors = {
    "Accessories": "tab:blue",
    "Footwear": "tab:orange",
    "Kidswear": "tab:green",
    "Menswear": "tab:red",
    "Womenswear": "tab:purple"
}

# =========================================================
# 13. CREATE FIGURE
# =========================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(18, 8)
)

# =========================================================
# 14. X POSITIONS
# =========================================================

x = np.arange(len(quarters))

# =========================================================
# 15. SALES CHART
# =========================================================

ax1 = axes[0]

bottom_sales = np.zeros(len(quarters))

for category in categories:

    values = sales[category].values

    bars = ax1.bar(
        x,
        values,
        bottom=bottom_sales,
        color=category_colors.get(
            category,
            "gray"
        ),
        width=0.65
    )

    # -----------------------------------------
    # SALES LABEL
    # -----------------------------------------

    for i, value in enumerate(values):

        if value > 0:

            ax1.text(
                x[i],
                bottom_sales[i] + value / 2,
                f"{value:,.0f}",
                ha="center",
                va="center",
                fontsize=8,
                fontweight="bold"
            )

    bottom_sales += values

# =========================================================
# 16. SALES CHART FORMAT
# =========================================================

ax1.set_title(
    "Quarterly Sales by Category",
    fontsize=15,
    fontweight="bold"
)

ax1.set_xlabel("Quarter")

ax1.set_ylabel("Sales Amount")

ax1.set_xticks(x)

ax1.set_xticklabels(
    quarters,
    rotation=0
)

ax1.grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

# =========================================================
# 17. RETURNS CHART
# =========================================================

ax2 = axes[1]

bottom_returns = np.zeros(len(quarters))

for category in categories:

    values = returns[category].values

    bars = ax2.bar(
        x,
        values,
        bottom=bottom_returns,
        color=category_colors.get(
            category,
            "gray"
        ),
        width=0.65,
        edgecolor="black"
    )

    # =====================================================
    # RETURN LABEL
    # =====================================================

    for i, value in enumerate(values):

        if value > 0:

            # If segment is large enough
            if value > 5000:

                ax2.text(
                    x[i],
                    bottom_returns[i] + value / 2,
                    f"{category}\n{value:,.0f}",
                    ha="center",
                    va="center",
                    fontsize=8,
                    fontweight="bold",
                    bbox=dict(
                        facecolor="white",
                        alpha=0.75,
                        edgecolor="none",
                        pad=2
                    )
                )

            # Small segment → label above bar
            else:

                ax2.annotate(
                    f"{category}: {value:,.0f}",
                    xy=(
                        x[i],
                        bottom_returns[i] + value
                    ),
                    xytext=(
                        x[i] + 0.18,
                        bottom_returns[i] + value + 8000
                    ),
                    fontsize=7,
                    fontweight="bold",
                    arrowprops=dict(
                        arrowstyle="->",
                        linewidth=0.8
                    ),
                    bbox=dict(
                        facecolor="white",
                        alpha=0.85,
                        edgecolor="black",
                        pad=2
                    )
                )

    bottom_returns += values

# =========================================================
# 18. RETURNS CHART FORMAT
# =========================================================

ax2.set_title(
    "Quarterly Returns (YES) by Category",
    fontsize=15,
    fontweight="bold"
)

ax2.set_xlabel("Quarter")

ax2.set_ylabel("Returned Sales Amount")

ax2.set_xticks(x)

ax2.set_xticklabels(
    quarters,
    rotation=0
)

ax2.grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

# =========================================================
# 19. COMMON CATEGORY LEGEND
# =========================================================

from matplotlib.patches import Patch

legend_items = []

for category in categories:

    legend_items.append(
        Patch(
            facecolor=category_colors.get(
                category,
                "gray"
            ),
            label=category
        )
    )

fig.legend(
    handles=legend_items,
    title="Category",
    loc="upper center",
    bbox_to_anchor=(0.5, 1.02),
    ncol=len(categories)
)

# =========================================================
# 20. MAIN TITLE
# =========================================================

fig.suptitle(
    "Quarterly Sales vs Returns by Category",
    fontsize=18,
    fontweight="bold"
)

# =========================================================
# 21. LAYOUT
# =========================================================

plt.tight_layout(
    rect=[0, 0, 1, 0.93]
)

plt.show()
```

```python

# RETURN RATE ANALYSIS 

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df["Return_Flag"] = (
    df["Return_Flag"]
    .astype(str)
    .str.strip()
    .str.upper()
)

print("\nReturn Flag Values:")
print(df["Return_Flag"].value_counts())

#  Calculate Total Orders
total_orders = len(df)
print("\nTotal Orders:", total_orders)

#  Calculate Returned Orders
returned_orders = (df["Return_Flag"] == "YES").sum()
print("Returned Orders:", returned_orders)

#  Calculate Return Rate KPI
return_rate = (returned_orders / total_orders) * 100

print("\n================================")
print("       RETURN RATE KPI")
print("================================")
print("Total Orders     :", total_orders)
print("Returned Orders  :", returned_orders)
print("Return Rate      : {:.2f}%".format(return_rate))
print("================================")

#  Create KPI DataFrame
kpi_df = pd.DataFrame({
    "KPI": [
        "Total Orders",
        "Returned Orders",
        "Return Rate (%)"
    ],
    "Value": [
        total_orders,
        returned_orders,
        round(return_rate, 2)
    ]
})

print("\nKPI Summary:")
print(kpi_df)
#  Month-wise Return Rate
# Convert Order_Date to datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Create Month column
df["Month"] = df["Order_Date"].dt.month_name()

# Create Month Number for proper sorting
df["Month_Number"] = df["Order_Date"].dt.month

month_analysis = (
    df.groupby(["Month_Number", "Month"])
    .agg(
        Total_Orders=("Return_Flag", "count"),
        Returned_Orders=("Return_Flag", lambda x: (x == "YES").sum())
    )
    .reset_index()
)

month_analysis["Return_Rate_%"] = (
    month_analysis["Returned_Orders"]
    / month_analysis["Total_Orders"]
    * 100
)

month_analysis = month_analysis.sort_values("Month_Number")

print("\nMonth-wise Return Rate:")
print(month_analysis)

# Category-wise Return Rate
category_analysis = (
    df.groupby("Category")
    .agg(
        Total_Orders=("Return_Flag", "count"),
        Returned_Orders=("Return_Flag", lambda x: (x == "YES").sum())
    )
    .reset_index()
)

category_analysis["Return_Rate_%"] = (
    category_analysis["Returned_Orders"]
    / category_analysis["Total_Orders"]
    * 100
)

print("\nCategory-wise Return Rate:")
print(category_analysis)

# Region-wise Return Rate
region_analysis = (
    df.groupby("Region")
    .agg(
        Total_Orders=("Return_Flag", "count"),
        Returned_Orders=("Return_Flag", lambda x: (x == "YES").sum())
    )
    .reset_index()
)

region_analysis["Return_Rate_%"] = (
    region_analysis["Returned_Orders"]
    / region_analysis["Total_Orders"]
    * 100
)

print("\nRegion-wise Return Rate:")
print(region_analysis)

#  Customer Segment-wise Return Rate
segment_analysis = (
    df.groupby("Customer_Segment")
    .agg(
        Total_Orders=("Return_Flag", "count"),
        Returned_Orders=("Return_Flag", lambda x: (x == "YES").sum())
    )
    .reset_index()
)

segment_analysis["Return_Rate_%"] = (
    segment_analysis["Returned_Orders"]
    / segment_analysis["Total_Orders"]
    * 100
)

print("\nCustomer Segment-wise Return Rate:")
print(segment_analysis)

#  Payment Mode-wise Return Rate
payment_analysis = (
    df.groupby("Payment_Mode")
    .agg(
        Total_Orders=("Return_Flag", "count"),
        Returned_Orders=("Return_Flag", lambda x: (x == "YES").sum())
    )
    .reset_index()
)

payment_analysis["Return_Rate_%"] = (
    payment_analysis["Returned_Orders"]
    / payment_analysis["Total_Orders"]
    * 100
)

print("\nPayment Mode-wise Return Rate:")
print(payment_analysis)
#  Return Status Count
return_status = df["Return_Flag"].value_counts()

print("\nReturn Status:")
print(return_status)

#  Return Rate Visualization
plt.figure(figsize=(6, 5))
plt.bar(
    ["Returned", "Not Returned"],
    [
        returned_orders,
        total_orders - returned_orders
    ]
)

plt.title("Returned vs Not Returned Orders")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")

plt.show()
#  Month-wise Return Rate Chart
plt.figure(figsize=(10, 5))

plt.plot(
    month_analysis["Month"],
    month_analysis["Return_Rate_%"],
    marker="o"
)

plt.title("Month-wise Return Rate")
plt.xlabel("Month")
plt.ylabel("Return Rate (%)")

plt.xticks(rotation=45)

plt.grid(True)

plt.show()

#  Category-wise Return Rate Chart
plt.figure(figsize=(8, 5))

plt.bar(
    category_analysis["Category"],
    category_analysis["Return_Rate_%"]
)

plt.title("Category-wise Return Rate")
plt.xlabel("Category")
plt.ylabel("Return Rate (%)")

plt.xticks(rotation=45)

plt.show()
#  Region-wise Return Rate Chart
plt.figure(figsize=(8, 5))

plt.bar(
    region_analysis["Region"],
    region_analysis["Return_Rate_%"]
)

plt.title("Region-wise Return Rate")
plt.xlabel("Region")
plt.ylabel("Return Rate (%)")

plt.xticks(rotation=45)

plt.show()

#  Find Highest Return Category
highest_category = category_analysis.loc[
    category_analysis["Return_Rate_%"].idxmax()
]

print("\nHighest Return Rate Category:")
print(highest_category)

# Find Highest Return Region
highest_region = region_analysis.loc[
    region_analysis["Return_Rate_%"].idxmax()
]

print("\nHighest Return Rate Region:")
print(highest_region)

# Final KPI Result

print("\n\n==========================================")
print("          FINAL RETURN RATE KPI")
print("==========================================")
print(f"Total Orders       : {total_orders}")
print(f"Returned Orders    : {returned_orders}")
print(f"Return Rate        : {return_rate:.2f}%")
print("==========================================")
```
